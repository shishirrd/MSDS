## `lists` vs `tuples`

Python has four collection types: Lists, Tuples, Dictionaries and Sets. For now we will focus primarily on lists.

1. **Lists** are ordered sequences of elements, with that order being specified by the order that the elements are in when the list is created or as elements are added to the list.  

    1. Lists are created using the `[]` syntax.
    
    2. Lists are <font color ='green'>**mutable**</font>. You can add, remove, and replace values using functions such as `append()`, `extend()`, `insert()`, `pop()`, `remove()`, and `del()`. 
    
    3. Lists can be created by string methods such as `split()` and `strip()`.
    
2. **Tuples** are similar to lists except for the very important fact that they are <font color = 'green'>**immutable**</font>.

    1. Tuples are created using the `()` syntax.
    
    2. Since Tuples are immutable, there are no functions that are built-in to modify the variables of Tuples.
    
    3. When to use a Tuple?  When you have data that will never change, like the days of the week:
       `days_of_the_week = ("Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday")`
    
3. **Both Lists and Tuples**

    1. Can include mixed data types.
    
    2. Are accessed by index.
    
    3. Contain a sequence of individual elements.
    
    4. Are stored in the order in that they were added.

### NumPy arrays vs. Lists

NumPy is a package that is used with Python for scientific computing. NumPy n-dimensional arrays are very convenient when working with large amounts of data. For more information on NumPy, refer to this resource: __[What is NumPy?](https://numpy.org/doc/2.2/user/whatisnumpy.html)__

Let's begin by creating a very simple 1D NumPy array, which looks a lot like a Python list.

````python
import numpy as np

my_arr = np.array([1,2,3])
type(my_arr)
my_arr
````

````python
# A 2D array
my_arr2 = np.array([(1,2,3),(4,5,6)])
my_arr2
````

Some things we can do with nparrays are very similar to lists.

````python
lst1 = [-3,2,19,10,-18,100,6]
lst2 = [0,1,2,3,4,5,6]
arr1 = np.array([-3,2,19,10,-18,100,6])
arr2 = np.array([0,1,2,3,4,5,6])

max(lst1)
np.max(arr1)
````

However, some things are not the same between the two. If we try to add two lists, Python will concatenate them. If we add two nparrays, Python will calculate the sum elementwise.

````python
lst1 + lst2
arr1 + arr2
````

Multiplying a list by an integer simply creates a new list with duplicates of the original. Whereas nparrays will multiply each element by the integer.

````python
lst = [1,2,3]
arr = np.array([1,2,3])

lst*3
arr*3
````

Lists can be converted to nparrays which can be convenient. Let's convert the `change_2020` list to a 1D nparray.

````python
arr_2020 = np.array(change_2020)
type(arr_2020)
````
